```python
import pandas as pd
df = pd.read_csv("employees.csv")

print("DATASET")
print(df)

print("\nSHAPE")
print(df.shape)

print("\nCOLUMNS")
print(df.columns)

print("\nFIRST 5 ROWS")
print(df.head())

print("\nLAST 5 ROWS")
print(df.tail())

print("\nINFO")
df.info()

print("\nSTATISTICS")
print(df.describe())
```

```python
print("Names:")
print(df["Name"])

print("\nSalaries:")
print(df["Salary"])

print("\nName and Salary:")
print(df[["Name", "Salary"]])

print("\nDepartment and Experience:")
print(df[["Department", "Experience"]])
```

```python
print("Salary greater than 80000 \n",df[df["Salary"] > 80000])
print("Employees Work in the finance \n", df[df["Department"] == "Finance"])
print("Employees experience greater than 5 or equal \n ", df[df["Experience"]>=5])
print("Employees from IT and Salary > 60000 \n",df[(df["Department"] == "IT") & (df["Salary"] > 60000)] )
print("Employees Salary <60000 or above >90000 \n",df[(df["Salary"] < 60000) | (df["Salary"] > 90000)])
```

```python
print(df[df["Salary"] > 70000][["Name","Salary"]])
print(df[df["Department"] == "IT"][["Name","Experience"]])
print(df[df["Experience"]>=5][["Name","Department","Salary"]])
print(df.iloc[0:3,[1,3]])
```

```python
print(df.sort_values("Salary"))
print(df.sort_values("Salary", ascending=False))
print(df.sort_values("Experience", ascending=False))
highest_paid = df.sort_values(
    "Salary",
    ascending=False
).head(1)

print(highest_paid)
highest_paid1 = df.sort_values(
    "Salary",
    ascending=False
)
print(highest_paid1.iloc[0:3])
print(df.sort_values("Experience",ascending=False).iloc[0:2])
print(df.sort_values("Department",ascending=False))
print(df.sort_values("Salary",ascending=False)[["Name","Salary"]])
```
