```python
import pandas as pd
salary = pd.Series([50000, 60000, 75000, 90000])
print(salary)
print(type(salary))
print(salary.shape)
print(salary.size)
print(salary.ndim)
```

```python
data = {
    "Employee_ID": [101, 102, 103, 104],
    "Name": ["Amit", "Rahul", "Priya", "Sneha"],
    "Salary": [50000, 75000, 60000, 90000]
}
df = pd.DataFrame(data)
print(df)
print(type(df))
print(df.shape)
print(df.size)
print(df.ndim)
```

```python
df.head()
df.head(2)
df.tail()
df.tail(2)
df.info()
df.describe()
```

```python
salary = df["Salary"]
print(salary)
print(type(salary))
print(salary.shape)
print(salary.mean())
print(salary.max())
print(salary.min())
```

```python
print(df.iloc[0])
print(df.iloc[1])
print(df.iloc[[0,2]])

print(df.iloc[0,2])
print(df.iloc[2,1])

print(df.loc[:, "Salary"])
print(df.loc[0, "Salary"])
```

```python
print(df[df["Salary"] > 60000])
print(df[(df["Salary"] >= 50000) & (df["Salary"] <= 80000)])
print(df[(df["Salary"] > 60000) & (df["Employee_ID"] < 104)])
print(df[(df["Salary"] < 60000) | (df["Salary"] > 80000)])
```
