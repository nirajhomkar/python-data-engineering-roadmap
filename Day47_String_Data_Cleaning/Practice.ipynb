```python
import pandas as pd
df = pd.DataFrame({
    "name": [" Rahul ", "PRIYA", " amit ", "Sneha"],
    "department": [" it ", "HR", "FINANCE", " It "]
})
df["name"] = df["name"].str.strip()
df["name"] = df["name"].str.title()
df["department"] = df["department"].str.strip()
df["department"] = df["department"].str.upper()
print(df)
```

```python
df = pd.DataFrame({
    "full_name": [
        "Rahul Homkar",
        "Priya Sharma",
        "Amit Kumar",
        "Sneha Patil"
    ]
})
df["first_name"] = df["full_name"].str.split(" ").str[0]
df["last_name"] = df["full_name"].str.split(" ").str[1]
print(df)
```

```python
df = pd.DataFrame({
    "name": [" Rahul ", "PRIYA", " Amit ", "Sneha", "RAJ"],
    "email": [
        "rahul@gmail.com",
        "priya@yahoo.com",
        "amit@gmail.com",
        "sneha@outlook.com",
        "raj@gmail.com"
    ],
    "department": [" IT ", "hr", "FINANCE", " It", "HR "]
})
df["name"] = df["name"].str.strip().str.title()
df["department"] = df["department"].str.strip().str.upper()
df["is_gmail"] = df["email"].str.endswith("@gmail.com")
print(df)
gmail_employees = df[df["is_gmail"]]
print(gmail_employees)
```

```python
import numpy as np
df = pd.DataFrame({
    "employee_id": [101, 102, 103, 101, 104, 105],
    "name": [" Rahul ", "PRIYA", " Amit ", "Rahul", "Sneha", "RAJ"],
    "email": [
        "rahul@gmail.com",
        "priya@yahoo.com",
        "amit@gmail.com",
        "rahul@gmail.com",
        "sneha@outlook.com",
        "raj@gmail.com"
    ],
    "department": [" IT ", "hr", "FINANCE", " IT", "HR ", "it"],
    "salary": [50000, 60000, None, 55000, 65000, None]
})
df["name"] = df["name"].str.strip().str.title()
df["department"] = df["department"].str.strip().str.upper()
df["is_gmail"] = df["email"].str.endswith("@gmail.com")
df["salary"] = df["salary"].fillna(df["salary"].median())
df = df.drop_duplicates(
    subset=["employee_id"],
    keep="last"
)
df["annual_bonus"] = df["salary"] * 0.10
df["salary_category"] = np.where(df["salary"] >= 60000,"High","Normal")
df["gmail_employees"] = df["email"].str.endswith("@gmail.com")
gmail_employees = df[df["is_gmail"]]
print(gmail_employees.isnull().sum())
print(gmail_employees.duplicated(subset=["employee_id"]).sum())
```
