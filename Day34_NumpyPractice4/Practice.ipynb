```python
import numpy as np
arr = np.array([12,24,36,48,60])
print(arr > 30)

print(arr < 40)

print(arr == 48)

print(arr != 24)
```

```python
import numpy as np

arr = np.array([8,16,24,32,40,48])

print("Greater than 20:", arr[arr > 20])

print("Less than or equal to 32:", arr[arr <= 32])

print("Equal to 40:", arr[arr == 40])

print("Not Equal to 16:", arr[arr != 16])
```

```python
import numpy as np
arr = np.array([12,18,24,30,36,42,48])
print("Between 20 and 40:", arr[(arr > 20) & (arr < 40)])
print("Less than 18 OR Greater than 40:",arr[(arr < 18) | (arr > 40)])
```

```python
import numpy as np
arr = np.array([25,50,75,100,125])
print("Not Greater than 75:", arr[~(arr > 75)])
print("Not equal to 50:", arr[~(arr == 50)])
```
