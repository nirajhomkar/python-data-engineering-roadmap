```python
import numpy as np
arr = np.array([65, 42, 89, 15, 73])
print(arr)
print("Minimum:", arr.min())

print("Maximum:", arr.max())

print("Sum:", arr.sum())

print("Mean:", arr.mean())
```

```python
import numpy as np
arr = np.array([10,20,30,40,50])
print(arr)

print("Sum:", arr.sum())

print("Mean:", arr.mean())

print("Minimum:", arr.min())

print("Maximum:", arr.max())

print("Standard Deviation:", arr.std())

print("Variance:", arr.var())
```

```python
import numpy as np
arr =  np.array([
    [5,10,15],
    [20,25,30],
    [35,40,45]
])
print(arr)

print("Sum Axis 0:", arr.sum(axis=0))

print("Sum Axis 1:", arr.sum(axis=1))

print("Mean Axis 0:", arr.mean(axis=0))

print("Mean Axis 1:", arr.mean(axis=1))
```
