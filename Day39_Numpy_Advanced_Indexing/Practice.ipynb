```python
import numpy as np

arr = np.array([10,20,30,40,50,60])

print("Original:")
print(arr)

print("Indexes 0, 2, 5:")
print(arr[[0,2,5]])

print("Indexes 1, 3, 4:")
print(arr[[1,3,4]])

indexes = np.array([0,3,5])

print(arr[indexes])
```

```python
import numpy as np

arr = np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])

print("Original:")
print(arr)

print("\nRows 0 and 2:")
print(arr[[0,2]])

print("\nRows 1 and 2:")
print(arr[[1,2]])
```

```python
import numpy as np

arr = np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])

rows = np.array([0,1,2])
cols = np.array([2,1,0])

print("Selected values:")
print(arr[rows, cols])
```

```python
import numpy as np

arr = np.array([10,20,30,40,50,60])

indexes = np.where(arr > 30)

print("Indexes:")
print(indexes)

print("Values:")
print(arr[indexes])
```

```python
import numpy as np

salaries = np.array([30000,45000,60000,75000,90000])

result = np.where(
    salaries >= 60000,
    "High",
    "Low"
)

print("Salaries:")
print(salaries)

print("Classification:")
print(result)

scores = np.array([45,67,82,91,38,76])

result = np.where(scores >= 50, "Pass", "Fail")

print(result)
```
