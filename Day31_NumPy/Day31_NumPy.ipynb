```python
import numpy as np
arr = np.array([
    [10,20,30],
    [40,50,60]
])
print(arr)
print(arr.shape)
print(arr.ndim)
print(arr.size)
print(arr.dtype)
```

```python
import numpy as np
arr = np.array([
    [100,200,300],
    [400,500,600]
])
print(arr[0,0])
print(arr[0,2])
print(arr[1,1])
print(arr[1,2])
```

```python
import numpy as np
arr = np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])
print(arr[0,:])
print(arr[:,2])
print(arr[1:3,:])
print(arr[:,0:2])
```
