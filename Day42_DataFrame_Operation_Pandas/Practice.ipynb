```python
import pandas as pd

data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Amit", "Rahul", "Priya", "Sneha", "Raj"],
    "Department": ["IT", "HR", "IT", "Finance", "IT"],
    "Salary": [50000, 75000, 60000, 90000, 55000]
}

df = pd.DataFrame(data)
df["Experience"] = [2, 4, 3, 6, 1]
df["Bonus"] = df["Salary"] * 0.10
print(df)
```

```python
tf = pd.DataFrame({
    "Employee_ID": [101, 102, 103],
    "Name": ["A", "B", "C"],
    "Salary": [40000, 60000, 80000],
    "Bonus": [4000, 6000, 8000]
})

tf = tf.rename(columns = {"Employee_ID":"ID"})
tf = tf.drop(columns = {"Bonus"})
tf["Salary"] = tf["Salary"] * 1.10
print(tf)
```

```python
print(tf)
print("\nData Types:")
print(tf.dtypes)
print("\nSalary type:")
print(tf["Salary"].dtype)
```

```python
df3 = pd.DataFrame({
    "Employee": ["A", "B", "C"],
    "Salary": ["45000", "60000", "75000"]
})
print(df3["Salary"].dtypes)
df3["Salary"] = pd.to_numeric(df3["Salary"])
print("Average Salary: ",df3["Salary"].mean())
print(df3["Salary"].dtypes)
```
