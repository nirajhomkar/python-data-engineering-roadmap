```python
import numpy as np

employees = np.array([
    [101, 50000, 2, 78],
    [102, 75000, 4, 91],
    [103, 45000, 1, 65],
    [104, 90000, 6, 95],
    [105, 60000, 3, 82],
    [106, 40000, 1, 58],
    [107, 85000, 5, 88],
    [108, 55000, 2, 72],
    [109, 95000, 7, 97],
    [110, 70000, 4, 85]
])
```

```python
print(employees)

total_employee = np.size(employees[:,0])
print(total_employee)
total_column = np.size(employees[0,:])
print(total_column)
total = np.shape(employees)
print(total)
print(employees[:,0])
print(employees[:,1])
print(employees[:,3])
salary = employees[:,1]
print(np.mean(salary))
print(np.max(salary))
print(np.min(salary))
performance = employees[:,3]
print(np.mean(performance))
```

```python
high_sal = employees[employees[:,1] > 70000]
print(high_sal)
```

```python
Performance_score = employees[employees[:,3] >= 85]
print(Performance_score)
```

```python
experience = employees[employees[:,2] >= 5]
print(experience)
```

```python
high_sal_perf = employees[(employees[:,1] > 70000) & (employees[:,3] >= 85)]
print(high_sal_perf)
```

```python
salary_status = np.where(salary >= 70000,"High","Low")
print(salary_status)
```

```python
print(np.sum(salary > 70000))
print(np.sum(employees[:,3] >= 85 ))
high_performers = employees[employees[:,3] >= 85]
high_performer_salaries = high_performers[:,1]
np.mean(high_performer_salaries)
```

```python
salary_status = np.where(
    salary >= 70000,
    "High",
    "Normal"
)
print(salary_status)
Performance_status = np.where(employees[:,3] >= 85, "Excellent", "Average")
print(Performance_status)
print(np.sum(Performance_status == "Excellent"))
maxi = np.max(employees[:,3])
print(maxi)
Best_Emp = np.where(employees[:,3] == maxi)
print(employees[Best_Emp])
```

```python
batches = np.array_split(employees, 3)
print(batches)
print("Number of batches:", len(batches))
```

```python
batch_report = []
for i, batch in enumerate(batches):
    employee_count = batch.shape[0]
    avg_salary = np.mean(batch[:,1])
    avg_performance = np.mean(batch[:,3])
    highest_performance = np.max(batch[:,3])

    print(f"\nBatch {i + 1}")
    print(f"Employees: {employee_count}")
    print(f"Average salary {avg_salary}")
    print(f"Average Performance: {avg_performance}")
    print(f"Highest Performance: {highest_performance}")

    row = [
        i+1,
        employee_count,
        avg_salary,
        avg_performance,
        highest_performance
    ]

    batch_report.append(row)
batch_report = np.array(batch_report)

print(batch_report)
print(batch_report.shape)
```

```python
max_avg_salary = np.max(batch_report[:,2])
print(max_avg_salary)

best_salary_batch = np.where(
    batch_report[:,2] == max_avg_salary
)

print(best_salary_batch)

highest_avg_perf = np.max(batch_report[:,3])
print(highest_avg_perf)

high_avg_batch = np.where(
    batch_report[:,3] == highest_avg_perf
)
print(high_avg_batch)

max_performance = np.max(employees[:,3])

best_employee_index = np.where(
    employees[:,3] == max_performance
)[0]

best_employee = employees[best_employee_index]

print(best_employee)

excellent_count = np.sum(employees[:,3] >= 85)

percentage = (
    excellent_count / employees.shape[0]
) * 100

print("Excellent employees:", excellent_count)
print("Percentage:", percentage)

overall_avg_salary = np.mean(employees[:,1])

overall_avg_performance = np.mean(employees[:,3])

print("===== FINAL EMPLOYEE ANALYTICS =====")

print("Overall Average Salary:", overall_avg_salary)
print("Overall Average Performance:", overall_avg_performance)
print("Excellent Employees:", excellent_count)
print("Excellent Employee Percentage:", percentage)
print("Best Employee:", best_employee)
```

```python
highest_avg_perf = np.max(batch_report[:,3])

high_avg_batch = np.where(
    batch_report[:,3] == highest_avg_perf
)[0]

print("Highest Average Performance:", highest_avg_perf)
print("Best Performance Batch:", high_avg_batch + 1)
```
