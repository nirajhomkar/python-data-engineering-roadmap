```python
import pandas as pd
df = pd.DataFrame({
    "name": ["Rahul", "Priya", "Amit", "Rahul", "Amit"],
    "salary": [50000, 60000, 70000, 50000, 70000],
    "department": ["IT", "HR", "IT", "IT", "IT"]
})
print(df)
print(df.duplicated())
print(df.duplicated().sum())
```

```python
clean_df = df.drop_duplicates()
print(clean_df)
print(clean_df.duplicated().sum())
```

```python
df = pd.DataFrame({
    "employee_id": [101, 102, 103, 101, 104, 103],
    "name": ["Rahul", "Priya", "Amit", "Rahul", "Sneha", "Amit"],
    "salary": [50000, 60000, 70000, 55000, 65000, 72000],
    "department": ["IT", "HR", "IT", "IT", "Finance", "IT"]
})
print(df.duplicated(subset=["employee_id"]))
print(df.duplicated(subset=["employee_id"]).sum())
clean_df = df.drop_duplicates(subset=["employee_id"],keep="last")
print(clean_df)
```

```python
df = pd.DataFrame({
    "employee_id": [101, 102, 103, 101, 104, 103, 105],
    "name": ["Rahul", "Priya", "Amit", "Rahul", "Sneha", "Amit", "Raj"],
    "salary": [50000, 60000, None, 55000, 65000, 72000, None],
    "department": ["IT", "HR", "IT", "IT", "Finance", "IT", None]
})
print(df.isnull().sum())
duplicate = df.duplicated(subset="employee_id",keep=False)
print(duplicate.sum())
clean_df = df.drop_duplicates(subset="employee_id",keep="last")
clean_df["salary"] = clean_df["salary"].fillna(clean_df["salary"].median())
clean_df["department"] = clean_df["department"].fillna("unknown")

print(clean_df)
print(clean_df.isnull().sum())
print(clean_df.duplicated(subset=["employee_id"]).sum())
```
