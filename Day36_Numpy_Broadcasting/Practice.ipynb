```python
import numpy as np

arr = np.array([10,20,30,40])

print("Original:", arr)

print("Add 5:", arr + 5)

print("Subtract 2:", arr - 2)

print("Multiply by 3:", arr * 3)

print("Divide by 2:", arr / 2)
```

```python
import numpy as np

arr = np.array([
    [10,20,30],
    [40,50,60]
])

print("Original:")
print(arr)

print("Add 5:")
print(arr + 5)

print("Multiply by 2:")
print(arr * 2)

print("Subtract 10:")
print(arr - 10)
```

```python
import numpy as np

arr = np.array([
    [10,20,30],
    [40,50,60]
])

values = np.array([1,2,3])

print("Array shape:", arr.shape)
print("Values shape:", values.shape)

print(arr + values)

values2 = np.array([1,2])

#print(arr + values2) gives broadcasting error
```

```python
import numpy as np

prices = np.array([
    [100],
    [200],
    [300]
])

tax = np.array([10,20])

print("Prices shape:", prices.shape)
print("Tax shape:", tax.shape)

print(prices + tax)
```
