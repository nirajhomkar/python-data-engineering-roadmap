```python
import pandas as pd
df = pd.DataFrame({
    "name": ["Rahul", "Priya", "Amit", "Sneha"],
    "salary": [50000, None, 60000, None],
    "department": ["IT", "HR", None, "IT"]
})
print(df)
print(df.isnull())
print(df.isnull().sum())
```

```python
print(df["salary"].notnull())
print(df[df["salary"].notnull()])
```

```python
print(df.dropna())
print(df)
```

```python
print(df["salary"].dropna())
```

```python
print(df.dropna(subset=["salary"]))
```

```python
print(df["salary"].fillna(0))
print(df["department"].fillna("unknown"))
print(df)
```

```python
df["salary"] = df["salary"].fillna(0)
df["department"] = df["department"].fillna("unknown")
print(df)
```

```python
df = pd.DataFrame({
    "name": ["Rahul", "Priya", "Amit", "Sneha"],
    "salary": [50000, None, 60000, None],
    "department": ["IT", "HR", None, "IT"]
})
print(df["salary"].median())
df["salary"]=df["salary"].fillna(df["salary"].median())
df["department"]=df["department"].fillna("unkown")
print(df)
print(df.isnull().sum())
```

```python
df = pd.DataFrame({
    "name": ["Rahul", "Priya", "Amit", "Sneha", "Raj"],
    "salary": [50000, None, 60000, None, 70000],
    "department": ["IT", "HR", None, "IT", None],
    "age": [25, None, 28, 30, None]
})

print(df.isnull().sum())
df["salary"] = df["salary"].fillna(df["salary"].median())
df["department"] = df["department"].fillna("unkown")
df["age"] = df["age"].fillna(df["age"].median())
print(df)
print(df.isnull().sum())
```
