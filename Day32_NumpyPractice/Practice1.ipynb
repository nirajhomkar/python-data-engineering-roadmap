```python
import numpy as np
arr1 = np.zeros(5)
arr2 = np.zeros((2,4))
print(arr1)
print(arr2)
print(arr1.shape)
print(arr2.shape)
```

```python
import numpy as np
arr1 = np.ones(6)
arr2 = np.ones((3,4))
print(arr1)
print(arr2)
print(arr1.shape)
print(arr2.shape)
```

```python
import numpy as np
arr1 = np.full(6,100)
arr2 = np.full((3,4), -5)
print(arr1)
print(arr2)
print(arr1.shape)
print(arr2.shape)
```

```python
import numpy as np
arr1 = np.arange(10)
arr2 = np.arange(5,15)
arr3 = np.arange(2,20,3)
print(arr1)
print(arr2)
print(arr3)
```

```python
import numpy as np
arr1 = np.linspace(0,20,5)
arr2 = np.linspace(10,100,10)
print(arr1)
print(arr2)
print(arr1.shape)
print(arr2.shape)
```

```python
import numpy as np
arr = np.arange(12)
print(arr)
arr2 = arr.reshape(4,3)
print(arr2)
print(arr2.shape)
```

```python
import numpy as np
arr1 = np.eye(3)
arr2 = np.eye(5)
print(arr1)
print(arr2)
print(arr1.shape)
print(arr2.shape)
```
