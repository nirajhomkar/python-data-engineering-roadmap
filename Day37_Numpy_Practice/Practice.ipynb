```python
import numpy as np
arr = np.arange(1,13)
print("Original:")
print(arr)

print("Original Shape")
print(arr.shape)

arr2 =  arr.reshape(3,4)

print("Reshaped:")
print(arr2)

print("New Shape:")
print(arr2.shape)

arr3 = arr.reshape(4,3)

print(arr3)

arr4 = arr.reshape(2,6)

print(arr4)
```

```python
import numpy as np

arr = np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])

print("Original:")
print(arr)

print("Original Shape:")
print(arr.shape)
flat = arr.flatten()

print("Flattened:")
print(flat)

print("Flattened Shape:")
print(flat.shape)

arr2 = np.array([
    [1,2],
    [3,4],
    [5,6]
])

print(arr2.flatten())
```

```python
import numpy as np

arr = np.array([
    [10,20],
    [30,40]
])

flat = arr.flatten()

flat[0] = 999

print("Original array:")
print(arr)

print("Flattened array:")
print(flat)

arr2 = np.array([
    [10,20],
    [30,40]
])

r = arr2.ravel()

r[0] = 999

print("Original array:")
print(arr2)

print("Ravel array:")
print(r)
```

```python
import numpy as np

arr = np.array([
    [10,20,30],
    [40,50,60]
])

print("Original:")
print(arr)

print("Original Shape:")
print(arr.shape)

transposed = arr.T

print("Transposed:")
print(transposed)

print("Transposed Shape:")
print(transposed.shape)

print(arr.transpose())
```

```python
import numpy as np
data = np.arange(1,13)
print("Original:")
print(data)
matrix = data.reshape(3,4)
print("\nReshaped:")
print(matrix)
print("\nTranspose:")
print(matrix.T)
print("\nFlatten:")
print(matrix.flatten())
print("\nRavel:")
print(matrix.ravel())
```
