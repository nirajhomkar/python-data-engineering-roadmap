```python
import pandas as pd
df = pd.DataFrame({
    "name": ["Rahul", "Priya", "Amit", "Sneha"],
    "salary": [50000, 60000, 70000, 80000]
})
df["annual_bonus"] = df["salary"] * 0.10
df["monthly_salary"] = df["salary"] / 12
print(df)
```

```python
df = pd.DataFrame({
    "name": ["Rahul", "Priya", "Amit", "Sneha"],
    "salary": [50000, 60000, 70000, 80000],
    "bonus": [5000, 6000, 7000, 8000],
    "tax": [5000, 6000, 7000, 8000]
})
df["total_compensation"] = df["salary"] + df["bonus"]
df["net_salary"] = df["salary"] + df["bonus"] - df["tax"]
print(df)
```

```python
import numpy as np 
df = pd.DataFrame({
    "name": ["Rahul", "Priya", "Amit", "Sneha", "Raj"],
    "salary": [45000, 55000, 70000, 85000, 95000]
})
df["select_category"] = np.where(df["salary"] >= 70000,"High","Normal")
print(df)
```

```python
df = pd.DataFrame({
    "name": ["Rahul", "Priya", "Amit", "Sneha", "Raj"],
    "salary": [45000, 55000, 70000, 85000, 95000]
})
def classify_salary(salary):
    if salary < 50000:
        return "Low"
    elif salary < 70000:
        return "Medium"
    elif salary < 90000:
        return "High"
    else:
        return "Very High"

df["salary_category"] = df["salary"].apply(classify_salary)
print(df)
```

```python
df = pd.DataFrame({
    "name": ["Rahul", "Priya", "Amit", "Sneha"],
    "salary": [50000, 60000, 70000, 80000]
})
df["salary_after_raise"] = df["salary"].apply(lambda salary : salary * 0.15)
print(df)
```

```python
df = pd.DataFrame({
    "name": ["Rahul", "Priya", "Amit", "Sneha", "Raj"],
    "department": ["IT", "HR", "Finance", "IT", "HR"]
})

department_map = {
    "IT": "Technology",
    "HR": "Human Resources",
    "Finance" : "Financial Services"
}

df["department_full"] = df["department"].map(department_map)
print(df)
```

```python
df = pd.DataFrame({
    "name": ["Rahul", "Priya", "Amit", "Sneha"],
    "salary": [45000, 60000, 75000, 90000],
    "department": ["IT", "HR", "Finance", "IT"]
})

df["bonus"] = df["salary"] * 0.10
df["total_compensation"] = df["salary"] + df["bonus"]
df["salary_category"] = np.where(
    df["salary"] >= 70000,
    "High",
    "Normal"
)
department_map = {
    "IT": "Technology",
    "HR": "Human Resources",
    "Finance": "Financial Services"
}

df["department_full"] = df["department"].map(department_map)
print(df)
```
